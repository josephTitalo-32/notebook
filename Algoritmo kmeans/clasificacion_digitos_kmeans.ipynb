# Clasificacion de Digitos con K-means
vamos a clasificar digitos de la base de datos contenida en la libreria sklearn de python. Utilizando el algoritmo K-means.

```python
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
from sklearn.cluster import KMeans
from sklearn.datasets import load_digits
```

Cargar dataset

```python
digitos=load_digits()
data=digitos.data
print(data.shape)
```

```python
data=255-data
np.random.seed(1)
n=10 #Numero de clusters o grupos
kmeans=KMeans(n_clusters=n, init="random")
kmeans.fit(data)
z=kmeans.predict(data)
```

Mostramos las imagenes segun la clasificacion del algoritmo K-Means

```python
for i in range(0,n):
    fila=np.where(z==i)[0] #filas en z en donde estan las imagens de cada cluster
    num=fila.shape[0] #numero de imagenes de cada cluster
    r=int(np.floor(num/10.)) #numero de filas mencionadas

    print ("cluster " + str(i))
    print(str(num) + " elementos")
    plt.figure(figsize=(10,10))
    for k in range(0,num):
        plt.subplot(r+1,10,k+1)
        imagen=data[fila[k],]
        imagen=imagen.reshape(8,8)
        plt.imshow(imagen,cmap=plt.cm.gray)
        plt.axis("off")
    plt.show()
```
