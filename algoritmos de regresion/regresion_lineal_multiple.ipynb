# Algoritmo de Regresion Multiple

```python
import pandas as pd
from sklearn.linear_model import LinearRegression
import numpy as np
from sklearn.metrics import r2_score
from mpl_toolkits.mplot3d import Axes3D
import matplotlib.pyplot as plt
```

Importamos los datos

```python
df=pd.read_excel("Datos/datos_producción.xlsx")
df.head()
```

Seleccionamos las variable

```python
x1="Horas Trabajadas"
x2="Horas Descanso"
y="Productos Terminados"
```

Crear resultados

```python
variables_x=[x1,x2]
variable_y=y

modelo=LinearRegression()
modelo.fit(df[variables_x], df[variable_y])#entrenamos el modelo

print("Coeficientes: ", modelo.coef_)
print("Intercepcion: ", modelo.intercept_)

print("ecuacion del plano: y={}* Horas Trabajadas + {} * Horas de descanso + {}".format(round(modelo.coef_[0],3) ,round(modelo.coef_[1],3), round(modelo.intercept_,3)))
```

Graficamos el plano en 3D

```python
fig=plt.figure()
ax=fig.add_subplot(111, projection="3d")
ax.scatter(df[x1], df[x2], df[y], color="blue")

x=np.linspace(df[x1].min(), df[x1].max(), num=10)
y=np.linspace(df[x2].min(), df[x2].max(), num=10)
x, y=np.meshgrid(x, y)

z=modelo.intercept_ + modelo.coef_[0]*x + modelo.coef_[1]*y

ax.plot_surface(x, y, z, alpha=0.5)
ax.set_xlabel(x1)
ax.set_ylabel(x2)
ax.set_zlabel("productos terminados")
plt.show()
```
