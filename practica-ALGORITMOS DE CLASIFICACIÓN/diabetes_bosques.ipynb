```python
import pandas as pd
import numpy as np
from sklearn.datasets import load_diabetes 
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, classification_report
import matplotlib.pyplot as plt
```

```python
#cargar datafrae
```

```python
data=load_diabetes()
x=pd.DataFrame(data.data, columns=data.feature_names)
y=(data.target>150).astype(int)
```

```python
x_train, x_test, y_train, y_test=train_test_split(x, y, test_size=0.3, random_state=42)
```

```python
#Entrenar el bosque aleatorio
```

```python
rf=RandomForestClassifier()
rf.fit(x_train,y_train)
```

```python
#Prediccionyevaluaciondelmodelov
```

```python
y_pred=rf.predict(x_test)
print(f"Exactitud: {accuracy_score(y_test, y_pred)}")
print(classification_report(y_test, y_pred))
```

```python
importances=rf.feature_importances_
indices=np.argsort(importances)[::-1]
plt.figure(figsize=(10,6))
plt.bar(range(x.shape[1]), importances[indices], align="center")
plt.xticks(range(x.shape[1]), x.columns[indices], rotation=90)
plt.title("importncia de las caracteristicas en ramdon forest")
plt.show()
```

```python
print("Graficar de la clasificacion de radom forest")
```

```python
from sklearn.metrics import ConfusionMatrixDisplay

plt.figure(figsize=(6, 6))
# Generar y mostrar la matriz de confusión directamente desde el modelo
ConfusionMatrixDisplay.from_estimator(
    rf, 
    x_test, 
    y_test, 
    display_labels=["Normal (<=150)", "Alto (>150)"], 
    cmap=plt.cm.Blues,
    values_format='d'
)
plt.title("Matriz de Confusión - Random Forest")
plt.grid(False) # Quitar la cuadrícula para que se vea más limpio
plt.show()
```
