```python
import seaborn as sns
import pandas as pd
from sklearn.linear_model import LinearRegression
import numpy as np
from sklearn.metrics import r2_score
import matplotlib.pyplot as plt
```

```python
#cargamos los datos
```

```python
df=pd.read_excel("Datos/datos_producción.xlsx")
df.head()
```

```python
#Seleccionar variables
```

```python
variable_x="Horas Trabajadas"
variable_y="Productos Terminados"
```

```python
#Generar analisis
```

```python
modelo=LinearRegression()
modelo.fit(df[[variable_x]], df[variable_y])

#Mostrar la ecuacion de la recta
print("Ecuacion de la recta: y=", round(modelo.coef_[0],3), "x + ", round(modelo.intercept_,3))
print("Coeficiente de correlacion: ", round(np.corrcoef(df[variable_x], df[variable_y])[0,1],3))
print("Coeficiente de determinacion: ", round(r2_score(df[variable_y], modelo.predict(df[[variable_x]])),3))
```

```python
#Graficamos con intervalos de confianza
```

```python
sns.regplot(x=df[variable_x], y=df[variable_y], ci=95, line_kws={"color":"red"}, scatter_kws={"color":"black"})

plt.title("Regresion lineal simple con intervalo de confianza al 95%")
```

```python
#Generar prediccion
```

```python
dato_predictor=15
horas_trabajadas_nuevas=pd.DataFrame([dato_predictor], columns=[variable_x])
autos_producidos=modelo.predict(horas_trabajadas_nuevas)
print("la prediccion de autos producidos para", horas_trabajadas_nuevas.iloc[0],[0],"horas trabajadas es", round(autos_producidos[0],2))
```
